
# Valorant Agent Popularity and Performance Across Competitive Ranks

## 1. Overview & Research Questions

This dataset research explores whether popular Valorant agents have higher win rates and whether this relationship changes across competitive ranks.

In my last proposal, I planned to compare agent pick rates and win rates across all players. However, since there are only around 28 agents and win rates are often close to 50%, I decided to expand my research by comparing multiple competitive ranks.

I am especially interested in Initiator because these agents are designed to help with duelists and control how the team's strategy on winning or losing. As my own experiences, a good duelist might only have more kills with a good initiator. I want to find out whether popular Initiator perform differently in lower ranks compared with higher ranks.

### Research Questions

1. What's the relationship between agent pick rate and win rate change across competitive ranks?
2. Do popular initiator agents have lower win rates in lower ranks than in higher ranks?
3. What's the differences in popularity and performance between agent roles across ranks?

### Hypotheses

H1: In lower ranks (Iron to Gold), agents with higher pick rates are more likely to have higher win rates. In higher ranks (Platinum to Immortal), the relationship between pick rate and win rate will be weaker.

H2: Popular Initiator agents will have lower win rates in lower ranks (Iron to Gold) compared with higher ranks (Platinum to Immortal).

H3: Different agent roles (Duelists, Controllers, Initiators, and Sentinels) will show different patterns in pick rates and win rates across the eight competitive ranks.

## 2. Background / Related Work

Valorant is a competitive shooting game where players choose agents with different abilities. Agents belong to four roles: Duelist, Controller, Initiator, and Sentinel. Each role has different responsibilities in a team. According to Riot Games, competitive ranks represent different player skill levels. This makes rank an important factor to consider when analyzing agent performance. Valorant Tracker provides agent statistics such as pick rates and win rates. These data can help show which agents are popular and how often they win. However, popularity does not necessarily represent performance because players may choose agents based on personal preference or playstyle. Research by Do et al. (2021) on League of Legends also shows that a player's experience with a particular character can influence game outcomes. Although League of Legends is a different game, this research suggests that player experience should be considered when interpreting character win rates.

By comparing different ranks, I hope to better understand whether the relationship between agent popularity and performance depends on player skill.

## 3. Methodology

### Data Collection

I plan to use Valorant Tracker to collect agent statistics from eight competitive ranks: Iron, Bronze, Silver, Gold, Platinum, Diamond, Ascendant, and Immortal.

For each rank, I will collect the following information:

- Agent name
- Agent role (Duelist, Controller, Initiator, or Sentinel)
- Pick rate (Played %)
- Win rate (Win %)
- Competitive rank
- Date of data collection

I will use the same map, region, and game mode settings for all eight ranks to make the results more consistent. I will record the filters and collection date for each dataset. The data will be saved in a CSV file and analyzed using Python.

### Data Cleaning

I will use Python and Pandas to organize and clean the data. I will check for missing values, duplicate records, and inconsistent agent names. I will also convert pick rates and win rates into numerical values for analysis.

### Data Analysis

**Research Question 1: Relationship Between Pick Rate and Win Rate Across Ranks**

I will create scatter plots showing the relationship between pick rate and win rate for each of the eight ranks. Each point will represent one agent.

I will use Pearson correlation to measure whether agents with higher pick rates tend to have higher win rates. I will then compare the results between lower ranks (Iron to Gold) and higher ranks (Platinum to Immortal) to see whether the relationship changes.

**Research Question 2: Initiator Performance Across Ranks**

I will focus on Initiator agents such as Sova, Fade, Skye, and Breach. I will compare their pick rates and win rates across the eight ranks.

I will create line charts to show how each Initiator's win rate changes as rank increases. I will also compare the most popular Initiators in lower ranks with their performance in higher ranks.

This will help me explore whether Initiators perform better at higher skill levels, where players may have stronger communication and teamwork.

**Research Question 3: Differences Between Agent Roles**

I will group agents into four roles: Duelist, Controller, Initiator, and Sentinel.

I will calculate the average pick rate and win rate for each role at each rank. I will use bar charts to compare these results and identify whether some roles are more popular or perform better at certain ranks.

### Presenting Findings

I plan to use data visualizations for presentation:

1. **Scatter plots:** Show the relationship between pick rate and win rate at different ranks.
2. **Line charts:** Show changes in Initiator win rates from Iron to Immortal.
3. **Bar charts:** Compare average pick rates and win rates between the four agent roles.

### Limitations

One limitation is that I haven't found a way to download the data directly from Tracker.gg. I may need to record the statistics manually, which could take a lot of time and increase the chance of human errors.

Another limitation is that win rates are often close to 50%. Even when comparing eight different ranks, the differences may still be small, making it difficult to identify clear patterns or meaningful relationships between pick rates and win rates.

Also, win rates can be affected by other factors, such as player skill, teamwork, map selection, and team composition. Since my dataset does not directly measure these factors, I cannot determine whether agent popularity is the reason for differences in win rates.
