
# HCDE 410 Final Project Proposal
## Valorant Agent Pick Rates vs. Win Rates

### 1. Motivation & Problem

My research question is: **Do the most popular Valorant agents have higher win rates?**

I chose this topic because I play Valorant and am interested in how players choose their agents. Some agents are picked more often than others, but I wonder if popular agents actually perform better.

Through this project, I want to compare agent pick rates and win rates to see if there is a relationship between them. I also want to explore whether different agent roles (Duelist, Controller, Initiator, and Sentinel) show different patterns.

I hope this project can help me better understand how game statistics relate to player choices and performance.

### 2. Data

My main data source is [Valorant Tracker](https://tracker.gg/valorant/insights/agents).

This website provides statistics about Valorant agents, including:
- Agent names and roles
- Pick rates
- Win rates
- Performance on different maps

I plan to focus on ranked competitive matches and collect statistics for all agents using the same region, rank, and time period.

I will collect the publicly displayed data manually and use Python to compare pick rates and win rates. I will create graphs to look for patterns.

**Data license and ethics:** Valorant Tracker is a third-party website, and I have not confirmed an open license for its statistics. I will check its terms before reusing or publishing the data. I will only use overall agent statistics, not individual players' personal information. I also understand that win rates can be affected by player skill, maps, and team compositions.

### 3. Unknowns & Dependencies

One challenge is that Valorant Tracker updates its statistics regularly, so the data may change during my project. I plan to record the date when I collect the data.

Another challenge is that since it's not all the users in valorant use Valorant Tracker, I may not get 100% of the data.
